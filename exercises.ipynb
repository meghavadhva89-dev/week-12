The [Knapsack Problem](https://en.wikipedia.org/wiki/Knapsack_problem) can be defined as follows:

*Given a set of items, each with a weight and a value, determine which items to include in the collection so that (a) the total weight is less than or equal to a given limit, and (b) the total value is as large as possible.*

Below is one particular solution to this problem. For this exercise,

1. In your own words, describe what this algorithm is doing.
2. Add a comment above *every* line, explaining what that line is doing (in reference to your explanation from #1).

**Algorithm Explanation (in my own words)**

This implementation uses dynamic programming to build up a table of optimal values.
Each row corresponds to considering items up to a given index, and each column corresponds to a capacity from 0 up to the limit W.
For each item and capacity, the algorithm decides whether including the current item (if it fits) yields a better total value than excluding it, and stores the best of the two choices in the table.
At the end, the entry in the bottom-right of the table contains the maximum value achievable for capacity W using all items.

```python
# Define the knapsack dynamic programming function with an option to return the full table
def knapsack(W, weights, values, full_table=False):
# Determine the number of items
    n = len(values)
# Initialize a (n+1) x (W+1) table filled with zeros
    table = [[0 for x in range(W + 1)] for x in range(n + 1)]

# Loop over all items (including the zero-th dummy row)
    for i in range(n + 1):
# Loop over all weight capacities from 0..W
        for j in range(W + 1):

# If we're in the first row or first column, the value is zero (no items or zero capacity)
            if i == 0 or j == 0:
                table[i][j] = 0

# If the current item's weight fits in capacity j, consider including it
            elif weights[i-1] <= j:
# Value if we include the current item (value of item + best for remaining capacity)
                a_1 = values[i-1]
# Remaining capacity after including the item
                diff = j - weights[i-1]
# Best value achievable with the remaining capacity using previous items
                a_2 = table[i-1][diff]
# Total value if we include the current item
                a = a_1 + a_2
# Value if we do NOT include the current item (best from previous items for same capacity)
                b = table[i-1][j]
# Choose the better of including or excluding the item
                table[i][j] = max(a, b)

# If the item doesn't fit, inherit the value from the previous row (don't include it)
            else:
                table[i][j] = table[i-1][j]

# If requested, return the full DP table for inspection
    if full_table:
        return table

# Otherwise return the optimal value for capacity W using all n items
    return table[n][W]
```
